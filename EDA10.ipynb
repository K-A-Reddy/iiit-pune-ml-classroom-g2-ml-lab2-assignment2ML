```python
#downloading the dataset
import pandas as pd
true_news= pd.read_csv("/Users/kosireddyanirudhreddy/Downloads/News _dataset/True.csv")
fake_news= pd.read_csv("/Users/kosireddyanirudhreddy/Downloads/News _dataset/Fake.csv")
```

```python
#inspecting the datasets to concatenate
true_news.head
true_news.columns
fake_news.columns
print(true_news.shape)
fake_news.shape
```

```python
#concatenating the datasets
true_news['label']=0
fake_news['label']=1
df=pd.concat([true_news,fake_news], axis=0)
df=df.sample(frac=1,random_state=42).reset_index(drop=True)
print(df.shape)
```

```python
from sklearn.model_selection import train_test_split
X=df[['title','text','subject','date']]
y=df['label']
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.1, random_state=42, stratify=y)
X1_train=X_train['text']
X1_test=X_test['text']
```

```python
# encoding the dataset using TF-IDF
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf=TfidfVectorizer(lowercase=True, stop_words='english')
X_train_tfidf= tfidf.fit_transform(X1_train)
X_test_tfidf= tfidf.transform(X1_test)
```

```python
# inspecting the new data
print(X_train_tfidf.shape)
X_train_tfidf.nnz / (X_train_tfidf.shape[0] * X_train_tfidf.shape[1])
```

```python
#training the machine
from sklearn.linear_model import LogisticRegression
log_reg= LogisticRegression(max_iter=1000, random_state=42)
log_reg.fit(X_train_tfidf,y_train)
y_pred=log_reg.predict(X_test_tfidf)
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
print("Accuracy :", accuracy_score(y_test, y_pred))
print("precision_score :", precision_score(y_test, y_pred))
print("recall_score :", recall_score(y_test, y_pred))
print("f1_score :" , f1_score(y_test,y_pred))
```

```python
# confusion matrix
from sklearn.metrics import confusion_matrix
confusion_matrix(y_test,y_pred)
```

```python
# removing the token "reuters"
leakage_tokens = ["reuters"]
def remove_leakage(text):
    if not isinstance(text, str):
        return text
    for token in leakage_tokens:
        text = text.replace(token, "")
        text = text.replace(token.capitalize(), "")
        text = text.replace(token.upper(), "")
    return text
X2_train=X1_train.apply(remove_leakage)
X2_test= X1_test.apply(remove_leakage)
tfidf_clean = TfidfVectorizer(
    lowercase=False,
    stop_words="english"
)
X_train_tfidf2 = tfidf_clean.fit_transform(X2_train)
X_test_tfidf2 = tfidf_clean.transform(X2_test)
log_reg = LogisticRegression(
    max_iter=1000,
    random_state=42
)

log_reg.fit(X_train_tfidf2, y_train)
y_proba2 = log_reg.predict_proba(X_test_tfidf2)[:, 1]
y_pred2 = (y_proba2 >= 0.3)
print("Precision:", precision_score(y_test, y_pred2))
print("Recall:", recall_score(y_test, y_pred2))
print("Confusion matrix:")
print(confusion_matrix(y_test, y_pred2))
```
