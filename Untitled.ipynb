```python
X=np.array([ [10, 100, 5], [12, 110, 6], [11, 105, 7], [13, 500, 6], [14, 108, 8], [15,115, 9] ])
m = np.mean(X, axis=0)
st = np.std(X, axis=0, keepdims=True)
median = np.median(X, axis=0)
outliers = np.abs(X - m) > 2 * st
print(outliers)
X_clean = X.copy()
X_clean[outliers] = np.broadcast_to(median, X.shape)[outliers]
print(X_clean)
```

```python
df = pd.DataFrame({
    'ID': [101,102,103,101,104,102,105],
    'Name': ['A','B','C','A','D','B','E'],
    'Age': [20,21,22,20,23,21,24],
    'Score': [80,75,90,85,88,75,92]
})
duplicated_ids = df[df['ID'].duplicated(keep=False)]['ID'].unique()
print(duplicated_ids)
```

```python
duplicates = df[df['ID'].duplicated(keep=False)]

identical = duplicates.groupby('ID').apply(
    lambda x: x.drop(columns='ID').nunique().max() == 1
)
print(identical)
```

```python
df_sorted = df.sort_values('Score', ascending=False)
clean_df = df_sorted.drop_duplicates('ID', keep='first')
clean_df = (
    df.sort_values('Score', ascending=False)
      .drop_duplicates('ID', keep='first')
      .sort_values('ID')
      .reset_index(drop=True)
)

print(clean_df)
```

```python
removed = len(df) - len(clean_df)
print("Records removed:", removed)
```

```python
df = pd.DataFrame({
    'Date': [
        '2026-01-01', '2026-01-02', '2026-01-04',
        '2026-01-05', '2026-01-08', '2026-01-09'
    ],
    'Sales': [100, 120, 150, 130, 180, 200]
})
df['Date'] = pd.to_datetime(df['Date'])
full_dates = pd.date_range(
    df['Date'].min(),
    df['Date'].max(),
    freq='D'
)
missing_dates = full_dates.difference(df['Date'])
print("Missing dates:", missing_dates)
df = df.set_index('Date').reindex(full_dates)
df['Rolling_Avg'] = df['Sales'].rolling(3).mean()
print(df)
print("Maximum rolling average date:", df['Rolling_Avg'].idxmax())
```

```python
import numpy as np
P = np.array([
    [1, 2],
    [4, 6],
    [7, 3],
    [2, 8],
    [9, 5]
])
diff = P[:, np.newaxis, :] - P[np.newaxis, :, :]
# Euclidean distances
D = np.sqrt(np.sum(diff**2, axis=2))
print(D)
```

```python
i, j = np.unravel_index(np.argmax(D), D.shape)
print("Maximum distance:", D[i, j])
print("Point pair:", i, j)
```
