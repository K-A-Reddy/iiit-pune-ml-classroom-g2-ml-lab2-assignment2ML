```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("Mall_Customers.csv")
df.head()
```

```python
df.info()
'''the dataset has no missing values and all the features are numerical except for the gender column which is categorical.
 The target variable is spending score.'''
df.describe()
'''the data has to be engineered for apt analysis. 
 The features are customer id, gender, age, annual income, and spending score.
 The dataset contains 200 samples.'''
```

```python
# remove the customer id column as it is not useful for analysis
df.drop('CustomerID', axis=1, inplace=True)
# convert the gender column to numerical values
df['Gender'] = df['Gender'].map({'Male': 0, 'Female': 1})
```

```python
# now we will plot a histogram to visualize the distribution of the target variable, spending score.
plt.hist(df['Spending Score (1-100)'], bins=10, alpha=0.5)
plt.xlabel('Spending Score')
plt.ylabel('Frequency')
plt.title('Distribution of Spending Score')
plt.show()
```

this plot shows us that most people have their scores at around 50 to 60

```python
# now we will plot histogams to analyze the distribution of the features in the dataset.
plt.hist(df['Age'], bins=10, alpha=0.5)
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.title('Distribution of Age')
plt.show()
```

```python
plt.hist(df['Annual Income (k$)'], bins=10, alpha=0.5)
plt.xlabel('Annual Income (k$)')
plt.ylabel('Frequency')
plt.title('Distribution of Annual Income')
plt.show()
# the plot shows us that we have more youngsters in the distribution and the annual income is distributed normally with a peak around 60k$.
```

```python
# now we will try to visualize the correlation of each feature with the target variable, spending score. We will use a scatter plot to visualize the correlation between the features and the target variable.
plt.scatter(df['Age'], df['Spending Score (1-100)'])
plt.xlabel('Age')   
plt.ylabel('Spending Score (1-100)')
plt.title('Correlation between Age and Spending Score')
plt.show()
# this shows us that we have no relation between age and spending score
# although we can see that old people in the dataset donot have high spenfing score at all
```

```python
plt.scatter(df['Annual Income (k$)'], df['Spending Score (1-100)'])
plt.xlabel('Annual Income (k$)')
plt.ylabel('Spending Score (1-100)')    
plt.title('Correlation between Annual Income and Spending Score')
plt.show()
# this shows us that we have no relation between annual income and spending score
# but we encounter a fact that people from 50 to 60 all have a specific spending score
```
